Write a python Program for Bidirectional Associative Memory with two pairs of vectors.			

```python
import numpy as np
```

```python
x1=np.array([1,1,-1,-1])
y1=np.array([1,-1,1])
```

```python
x2=np.array([1,-1,1,-1])
y2=np.array([-1,1,-1])
```

```python
W=np.outer(x1,y1)+np.outer(x2,y2)
```

```python
W
```

```python
def af(x):
  return np.where(x>=0,1,-1)
```

```python
def forward(x):
  return af(np.dot(x,W))
```

```python
def backward(x):
  return af(np.dot(x,W.T))
```

```python
print("for",x1)
print("froward of x1",forward(x1))
print("backward of y1",backward(y1))
```

```python
print("for",x2)
print("froward of x2",forward(x2))
print("backward of y2",backward(y2))
```
