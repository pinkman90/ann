Write a Python program to plot a few activation functions that are being used in neural networks.			

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1/(1+np.exp(-x))

def tanh(x):
  return np.tanh(x)

def relu(x):
  return np.maximum(0,x)

def leaky_relu(x,c=0.01):
  return np.where(x>0,x,c*x)

def softmax(x):
  exp_x=np.exp(x-np.max(x))
  return exp_x/np.sum(exp_x)

x=np.linspace(-10,10,100)

plt.figure(figsize=(10,5))

plt.subplot(4,2,1)
plt.plot(x,sigmoid(x))
plt.title("sigmoid")
plt.grid(True)

plt.subplot(4,2,2)
plt.plot(x,tanh(x))
plt.title("tanh")
plt.grid(True)

plt.subplot(4,2,3)
plt.plot(x,relu(x))
plt.title("relu")
plt.grid(True)

plt.subplot(4,2,4)
plt.plot(x,leaky_relu(x))
plt.title("leaky relu")
plt.grid(True)

plt.subplot(4,2,5)
plt.plot(x,softmax(x))
plt.title("softmax")
plt.grid(True)

plt.tight_layout()
plt.show()
```
