Generate AND/OR function using McCulloch-Pitts neural net by a python program.			

```python
def mp(inputs,w,t):
    weightedsum = 0

    for i in range(len(inputs)):
        weightedsum += (inputs[i]*w[i])

    if weightedsum >= t:
        return 1
    else:
        return 0

inputs=[
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
]

print("and gate")
print("x1 x2 o/t")
for j in inputs:
    output=mp(j, [1,1], 2)
    print(j[0], j[1]," ",output)


print("--"*100)
print("OR Gate")
print("x1 x2 O/P")

for j in inputs:
    output = mp(j, [1, 1], 1)
    print(j[0], j[1], " ", output)    
```
