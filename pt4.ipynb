With a suitable example demonstrate the perceptron learning law with its decision regions using python. Give the output in graphical form			

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# AND Gate Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,0,0,1])


# Initial parameters
weights = np.zeros(2)
bias = 0

learning_rate = .1
epochs = 5

error_list = []


# Perceptron Learning
for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        
        z = np.dot(X[i], weights) + bias

       
        prediction = 1 if z >= 0 else 0

        
        error = y[i] - prediction

        
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error

        total_error += abs(error)

    error_list.append(total_error)

    print("Epoch", epoch+1, "Error =", total_error)





xx, yy = np.meshgrid(np.arange(-0.5, 1.5, 0.01),
                     np.arange(-0.5, 1.5, 0.01))
Z = (xx * weights[0] + yy * weights[1] + bias >= 0).astype(int)            
plt.contourf(xx, yy, Z, alpha=0.4)            
plt.scatter(X[:,0], X[:,1])


plt.title("AND Gate Decision Region")
plt.xlabel("X1")
plt.ylabel("X2")
plt.grid()
plt.show()

plt.plot(range(1, epochs+1), error_list, marker='o')
plt.xlabel("Epoch")
plt.ylabel("Error")
plt.title("Error vs Epoch")
plt.grid()
plt.show()

print("Weights:", weights)
print("Bias:", bias)
```
