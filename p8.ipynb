```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
x = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=42)
```

```python
# ONE - HOT encoding 
y_train = np.eye(3)[y_train]
```

```python
w1 = np.random.randn(4, 100)                                     # Input -> Hidden layer connection
b1 = np.zeros((1,100))

w2 = np.random.randn(100,3)                                      # Hidden -> output layer connection
b2 = np.zeros((1,3))
```

```python
def relu(x):
    return np.maximum(x,0) # For forwrd pass 

def relu_derivative(x):
    return (x>0).astype(float) # For Backward 

def softmax(x):
    exp = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp/np.sum(exp, axis = 1, keepdims=True)
```

```python
lr = 0.1
epochs = 1000
m = x_train.shape[0]
m
```

```python
for epoch in range(epochs):
    
    # FORWARD PASS
    z1 = np.dot(x_train, w1) + b1 # calculate neurons for input->hidden layer
    A1 = relu(z1) 

    z2 = np.dot(A1, w2) + b2 # For hidden->output layer
    A2 = softmax(z2)
    
    # Cross entropy loss
    proba = np.sum(y_train*A2, axis = 1)
    loss = -np.mean(np.log(proba + 1e-8))

    # BAKCWARD PASS
 
    dz2 = (A2 - y_train) / m # Calculate the error and devide by m to get average gradient

    dw2 = np.dot(A1.T, dz2) # Calculate gradient for output layer weights 
    db2 = np.sum(dz2, axis=0, keepdims=True) # Calculate gradient for output layer biases 

    dA1 = np.dot(dz2, w2.T) # Send output layer gradients back to hidden layer 

    dz1 = dA1 * relu_derivative(z1) # Apply relu derivative

    dw1 = np.dot(x_train.T, dz1) # Calculate gradient for input -> hidden layer weights 
    db1 = np.sum(dz1, axis=0, keepdims=True)  # Calculate gradient biases for input layer->hidden

    w1 -= lr*dw1
    b1 -= lr*db1

    w2 -= lr*dw2
    b2 -= lr*db2

    if epoch%100==0:
        print("epoch :", epoch, " Loss :", loss)
    
```

```python
# Testing
z1 = np.dot(x_test, w1) + b1
A1 = relu(z1)
z2 = np.dot(A1, w2) + b2
A2 = softmax(z2)

y_pred = np.argmax(A2, axis=1)
accu = np.mean(y_pred == y_test)*100
print(accu)
```

```python
smaple = np.array([[5.1, 3.5, 1.4, 0.2]])
sample = scaler.transform(smaple)
z1 = np.dot(sample, w1) + b1
A1 = relu(z1)
z2 = np.dot(A1, w2) + b2
A2 = softmax(z2)
pred = np.argmax(A2, axis=1)[0]

print("Predicted Class :", iris.target_names[pred])
```
